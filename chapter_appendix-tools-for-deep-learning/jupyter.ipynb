```python
# @title
#1. Dataset Preparation

from sklearn.datasets import fetch_openml
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = fetch_openml(data_id=42352, as_frame=True).frame

# Show dataset basics
print("Shape:", df.shape)
print("Columns:\n", df.columns)
print("\nFirst 5 rows:\n", df.head())
```

```python
# @title
print("\n===== CENTRAL TENDENCY =====")
cols = ['G1', 'G2', 'G3', 'studytime', 'absences']

mean_values = df[cols].mean()
median_values = df[cols].median()
mode_values = df[cols].mode()

print("Mean:\n", mean_values)
```

```python
# @title
print("\nMedian:\n", median_values)
```

```python
# @title
print("\nMode:\n", mode_values)
```

```python
# @title
print("\n===== DISPERSION CODING =====")

# Range
range_values = df[cols].max() - df[cols].min()

# Variance
variance_values = df[cols].var()

# Standard deviation
std_values = df[cols].std()

print("Range:\n", range_values)
```

```python
# @title
print("\nVariance:\n", variance_values)
```

```python
# @title
print("\nStandard Deviation:\n", std_values)
```

```python
# @title
print("\n===== DISTRIBUTION ANALYSIS =====")

skewness_values = df[cols].skew()
print(skewness_values)
```

```python
# @title
print("\n===== DATA VISUALIZATION =====")
print("\n===== A.Univariate Analysis =====")

# Histogram of final grades
sns.histplot(df['G3'], kde=True)
plt.title("Distribution of Final Grades (G3)")
plt.show()

# Boxplot
sns.boxplot(x=df['G3'])
plt.title("Boxplot of Final Grades")
plt.show()
```

```python
# @title
print("\n===== DATA VISUALIZATION =====")
print("\n===== B.Bivariate Analysis =====")

sns.scatterplot(x='studytime', y='G3', data=df)
plt.title("Study Time vs Final Grade")
plt.show()
```

```python
# @title
print("\n===== DATA VISUALIZATION =====")
print("\n===== C.Multivariate Analysis =====")

# Pairplot
sns.pairplot(df[['G1','G2','G3','studytime','absences']])
plt.show()

# Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df[['G1','G2','G3','studytime','absences']].corr(),
            annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
